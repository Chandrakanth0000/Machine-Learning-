```python
data = {
    "Name": ["Ravi", "Sai", "Kiran"],
    "Age": [20, 21, 19],
    "Marks": [85, 92, 76]
}

df = pd.DataFrame(data)

print(df)
```

```python
print(df.isnull().sum())
```

### 28. DROP MISSING VALUES

```python
df_without_missing = df.dropna()

print(df_without_missing)
```

```python
df["Marks"] = df["Marks"].fillna(
    df["Marks"].mean()
)
```

### 36. GROUPBY

```python
data = {
    "City": [
        "Vizag",
        "Vizag",
        "Hyderabad",
        "Hyderabad"
    ],
    "Marks": [
        80,
        90,
        70,
        85
    ]
}

df = pd.DataFrame(data)

print(
    df.groupby("City")["Marks"].mean()
)
```

### 37. NUMPY VS PANDAS

```python
import numpy as np

marks = np.array([
    80,
    90,
    75,
    95
])

print(np.mean(marks))

data = {
    "Name": [
        "Ravi",
        "Sai",
        "Kiran",
        "Anu"
    ],
    "Marks": [
        80,
        90,
        75,
        95
    ]
}

df = pd.DataFrame(data)

print(df["Marks"].mean())
```

### 38. STUDENT PERFORMANCE ANALYZER

```python
df = pd.read_csv("students.csv")

print(df.head())
```

#### Calculate Average

```python
df["Average"] = (
    df["Python"] +
    df["C"] +
    df["Java"]
) / 3

print(df)
```

#### Highest Performer

```python
top_student = df.loc[
    df["Average"].idxmax()
]

print(top_student)
```

#### Lowest Performer

```python
low_student = df.loc[
    df["Average"].idxmin()
]

print(low_student)
```

#### Pass / Fail

```python
df["Result"] = df["Average"].apply(
    lambda x: "Pass" if x >= 35 else "Fail"
)

print(df)
```

#### City-wise Average

```python
city_average = (
    df.groupby("City")["Average"].mean()
)

print(city_average)
```

#### Sort Students

```python
df = df.sort_values(
    "Average",
    ascending=False
)

print(df)
```

#### Final Report

```python
print("Total Students :", len(df))

print(
    "Class Average :",
    df["Average"].mean()
)

print(
    "Highest Average :",
    df["Average"].max()
)

print(
    "Lowest Average :",
    df["Average"].min()
)

print("\nStudent Report:")

print(df)
```
